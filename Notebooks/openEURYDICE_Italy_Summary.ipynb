## Step 5: System-Level Indicators (ELET) - Key Education Metrics

```python
# ELET System-level indicators
elet_file = eurydice_path / 'ELET_System-level_indicators_2024_2025_open_data.xlsx'
xls_elet = pd.ExcelFile(elet_file)

print(f"ELET Indicator Sheets: {xls_elet.sheet_names}")

# Load each indicator sheet
elet_indicators = {}
for sheet in xls_elet.sheet_names:
    if 'Indicator' in sheet and 'Policy' not in sheet:
        df = pd.read_excel(elet_file, sheet_name=sheet, header=None)
        elet_indicators[sheet] = df
        print(f"\n{sheet}: {df.shape}")
        print(df.head(15))
```

```python
print("="*70)
print("ITALY'S EDUCATION SYSTEM: REGIONAL VARIANTS IN INSTRUCTION TIME")
print("="*70)

# Load detailed info from each Italy sheet
italy_instruction_data = {}
for sheet in italy_sheets:
    df = pd.read_excel(instr_time_file, sheet_name=sheet)
    italy_instruction_data[sheet] = df

print(f"\nItaly has {len(italy_sheets)} education jurisdictions:")
for i, sheet in enumerate(italy_sheets, 1):
    print(f"  {i}. {sheet}")

# Try to extract key information from first Italy sheet
print("\n" + "="*70)
print("SAMPLE: Italy Instruction Time Questionnaire (IT_1)")
print("="*70)

# Find rows with data (not NaN in first column)
df_it1 = italy_instruction_data['IT_1']
print(f"\nShape of IT_1 data: {df_it1.shape}")
print("\nLooking for country information in IT_1...")

# Check for country reference
for idx, row in df_it1.iterrows():
    row_str = ' '.join(str(val) for val in row.values if pd.notna(val))
    if 'Country' in row_str or 'Italy' in row_str:
        print(f"Row {idx}: {row_str[:150]}...")
        break

print("\nDetailed IT_1 structure (rows 0-15):")
print(df_it1.iloc[0:15, 0:5].to_string())
```

```python
print("\n" + "="*70)
print("ITALY: SCHOOL YEAR AND ACADEMIC CALENDAR (2024/2025)")
print("="*70)

# Filter for Italy in school year data (rows where first column contains 'IT')
df_yearstart_it = df_yearstart[df_yearstart[0].astype(str).str.contains('IT', case=False, na=False)]
print("\nItaly entries in School Year data:")
print(df_yearstart_it.head(10))

print("\n" + "="*70)
print("ITALY: ACADEMIC YEAR START (2025/2026)")
print("="*70)

# Filter for Italy in academic overview
df_academic_it = df_academic[df_academic[0].astype(str).str.contains('Italy|IT', case=False, na=False)]
print("\nItaly entries in Academic Overview:")
print(df_academic_it.head(10))

# Try to find Italy more broadly
print("\nSearching all academic data for Italy...")
italy_found_rows = []
for idx, row in df_academic.iterrows():
    row_str = ' '.join(str(val) for val in row.values if pd.notna(val))
    if 'Italy' in row_str:
        italy_found_rows.append(idx)
        
if italy_found_rows:
    print(f"Found Italy mentions at rows: {italy_found_rows}")
    for idx in italy_found_rows[:5]:
        print(f"\nRow {idx}:")
        print(df_academic.iloc[idx].to_string()[:200])
else:
    print("Checking alternative names...")
    # Display all countries in academic data
    print("\nAll countries/regions in Academic Overview data:")
    countries = df_academic[0].dropna().unique()
    for country in countries[7:20]:  # Skip first few header rows
        print(f"  - {country}")
```

```python
print("\n" + "="*70)
print("ITALY: EQUITY INDICATORS ANALYSIS (2024/2025)")
print("="*70)

# Analyze equity indicators for Italy
equity_italy_data = {}
for sheet_name, df_equity in equity_indicators.items():
    # Find Italy in the data (looking for 'IT' in first column)
    italy_rows = df_equity[df_equity[0].astype(str).str.contains('IT', case=False, na=False)]
    if len(italy_rows) > 0:
        equity_italy_data[sheet_name] = italy_rows
        print(f"\n{sheet_name}:")
        print(f"  Found {len(italy_rows)} rows for Italy")
        # Show first row of Italy data
        if len(italy_rows) > 0:
            print(f"  Sample: {italy_rows.iloc[0].to_string()[:150]}...")

# Display all countries in equity data to understand structure
print("\n" + "-"*70)
print("Countries/regions in Equity Indicators:")
first_equity_sheet = list(equity_indicators.values())[0]
countries_in_equity = first_equity_sheet[0].dropna().unique()
print(f"Total unique entries in first column: {len(countries_in_equity)}")
# Show entries that might be country codes
country_codes = [c for c in countries_in_equity if isinstance(c, str) and len(c) <= 4 and c.isupper()]
print(f"Country codes found: {sorted(country_codes)[:10]}...")
```

   - Early warning systems in place
   - Targeted support for vulnerable groups

   - Managing grade repetition
