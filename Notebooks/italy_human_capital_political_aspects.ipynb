# Italy: Human Capital, Migration, and Political Trust

This notebook expands the scope of the NEET and education research to include macroeconomic and political consequences, specifically mapping the interplay between:
1. Youth unemployment and NEET rates.
2. The migration divide (equity in education outcomes).
3. Political aspects: Institutional trust and life satisfaction.
4. Human capital outcomes (Brain Drain vs. skill gaps).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
ROOT = Path.cwd().parents[0] if Path.cwd().name == 'Notebooks' else Path.cwd()
LOCAL = ROOT / 'local_data'
EUROSTAT = LOCAL / 'eurostat'
```

## 1. The Migration Equity Gap (Early School Leavers)
Migrant populations in Italy face disproportionate risk for early school leaving, creating a fractured human capital pipeline.

```python
esl_path = EUROSTAT / 'eurostat_esl_by_migration.csv'
if esl_path.exists():
    esl = pd.read_csv(esl_path, low_memory=False)
    # Filter for Italy, Total sex, specific age ranges if needed
    esl_it = esl[(esl['geo'] == 'IT') & (esl['sex'] == 'T') & (esl['age'] == 'Y18-24')].copy()
    if not esl_it.empty:
        display(esl_it.groupby('TIME_PERIOD')['OBS_VALUE'].mean())
else:
    print("ESL data not found.")
```

## 2. Youth Unemployment Trend
Unemployment acts as the core transitional failure post-education, feeding directly into the NEET pool.

```python
yu_path = EUROSTAT / 'estat_youth_unemployment_rate.csv'
if yu_path.exists():
    yu = pd.read_csv(yu_path)
    # TIPSLM80 - Youth unemployment rate
    yu_it = yu[(yu['geo'] == 'IT') & (yu['sex'] == 'T')].sort_values('TIME_PERIOD')
    
    if not yu_it.empty:
        plt.figure(figsize=(10, 4))
        sns.lineplot(data=yu_it, x='TIME_PERIOD', y='OBS_VALUE', marker='o')
        plt.title('Italy Youth Unemployment Rate (15-24)')
        plt.ylabel('% of active population')
        plt.xlabel('Year')
        plt.show()
else:
    print("Youth unemployment data not found.")
```

## 3. Political Proxies: Institutional Trust
High NEET and unemployment rates correlate strongly with deteriorating trust in political and civic institutions, especially among the most impacted socio-demographics.

```python
trust_path = EUROSTAT / 'eurostat_institutional_trust.csv'
if trust_path.exists():
    trust = pd.read_csv(trust_path, low_memory=False)
    trust_it = trust[(trust['geo'] == 'IT')].copy()
    
    if not trust_it.empty:
        display(trust_it[['TIME_PERIOD', 'OBS_VALUE']].dropna().head())
else:
    print("Trust data not found.")
```

## Summary Analysis

Italy faces a dual crisis of **Human Capital**:
1. **Internal Under-utilization**: Manifested in high NEET rates and structural youth unemployment. Significant equity gaps exist along **migration/citizenship** lines, where non-natives endure steep early school leaving disparities.
2. **Political Disenfranchisement**: A lack of robust educational integration feeds into lower institutional trust.
3. **Capital Flight**: As explored in supplementary analysis, the highly educated segment tends to emigrate (Brain Drain), leaving the domestic economy with reduced innovation capability, whilst failing to skill up its vulnerable and migrant youth.
