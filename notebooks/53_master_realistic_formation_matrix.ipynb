# Notebook 53: The Realistic Formation Matrix

In this notebook, we enrich the Master OED Destination Matrix with the specific daily reality of each track:
- **PCTO Hours:** The statutory unpaid labor mandate (Legge 145/2018).
- **Study Hours:** The weekly temporal burden.
- **Financial Burden:** The expected first-year textbook cost.


```python
import json
from pathlib import Path

ROOT = Path('c:/Users/Dell/Documents/Antigravity/Italienation').resolve()
PROC = ROOT / 'local_data/processed'
json_path = PROC / 'master_oed_destination_matrix.json'

with open(json_path, 'r', encoding='utf-8') as f:
    catalog = json.load(f)

def get_formation_reality(tipo):
    tipo = str(tipo).lower()
    if 'liceo' in tipo:
        return {
            "Statutory_PCTO_Hours_Triennio": 90,
            "Total_Weekly_Study_Hours": "27-30",
            "Avg_First_Year_Textbook_Cost_EUR": 335.0
        }
    elif 'tecnico' in tipo:
        return {
            "Statutory_PCTO_Hours_Triennio": 150,
            "Total_Weekly_Study_Hours": "32",
            "Avg_First_Year_Textbook_Cost_EUR": 260.0
        }
    else:
        return {
            "Statutory_PCTO_Hours_Triennio": 210,
            "Total_Weekly_Study_Hours": "32",
            "Avg_First_Year_Textbook_Cost_EUR": 190.0
        }

# Traverse and inject
for reg, provs in catalog.items():
    if "Provinces" not in provs: continue
    for prov, munis in provs["Provinces"].items():
        if "Municipalities" not in munis: continue
        for muni, tracks in munis["Municipalities"].items():
            if "Tracks" not in tracks: continue
            for tipo, subtracks in tracks["Tracks"].items():
                formation = get_formation_reality(tipo)
                for ind, data in subtracks.items():
                    # Inject Formation Reality
                    data["Formation_Reality"] = formation

# Export updated JSON
out_json_path = PROC / 'master_realistic_formation_matrix.json'
with open(out_json_path, 'w', encoding='utf-8') as f:
    json.dump(catalog, f, ensure_ascii=False, indent=2)

print(f"Master Realistic Formation JSON exported to {out_json_path}")
```
