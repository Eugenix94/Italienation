This notebook analyzes the **OED (Origin -> Education -> Destination)** social mobility model, directly mapping Italy's performance against the latest global frameworks (Eurostat/OECD 2025-2026).

We also generate comparative OED vectors for Germany and the UK to highlight the structural anomalies of the Italian Tripartite system.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs('local_data/processed/oed', exist_ok=True)
```

### 1. Eurostat 2025 NEET Rates: The Global Status Quo
According to the latest Eurostat 2025 data, the EU-wide average NEET rate (15-29) is **11.0%**.

Italy sits at **13.3%** in 2025, which remains structurally higher than the EU average and drastically higher than comprehensive systems.

```python
eurostat_2025 = {
    'Country': ['Netherlands', 'Sweden', 'Germany', 'EU Average', 'UK (Est)', 'Italy', 'Romania'],
    'System_Type': ['Tracking', 'Comprehensive', 'Dual VET', 'Mixed', 'Comprehensive', 'Tripartite', 'Mixed'],
    'NEET_Rate_2025': [5.3, 5.9, 8.8, 11.0, 12.6, 13.3, 19.2]
}
df_eurostat = pd.DataFrame(eurostat_2025)
df_eurostat = df_eurostat.sort_values(by='NEET_Rate_2025')
df_eurostat.to_csv('local_data/processed/oed/eurostat_2025_neet_comparison.csv', index=False)
df_eurostat
```

### 2. Comparative OED Triangles: Italy vs Germany vs UK
The Goldthorpe OED model identifies two critical vectors of social immobility:
1. **O $\rightarrow$ E (Tracking Bias):** How strictly social origin determines the education track.
2. **O $\rightarrow$ D (Direct Effect of Origin / Nepotism):** How strictly social origin determines destination (income/employment) independently of education.

We model these vectors comparatively.

```python
comparative_oed = {
    'System': ['Italy (Tripartite)', 'Germany (Dual VET)', 'UK (Comprehensive)'],
    'O_to_E_Bias_Strength': ['Very High (Early Tracking at 14)', 'High (Tracking at 10, but porous)', 'Low (Unified curriculum to 16)'],
    'O_to_D_Direct_Effect': ['Very High (Nepotism/Lavoro Nero)', 'Low (Strong institutional transitions)', 'Medium (Class-based networking)'],
    'E_to_D_Meritocracy': ['Weak (High Brain Drain/Mismatch)', 'Strong (Direct hiring via apprenticeships)', 'Medium (Degree inflation)']
}
df_comp_oed = pd.DataFrame(comparative_oed)
df_comp_oed.to_csv('local_data/processed/oed/comparative_oed_vectors.csv', index=False)
df_comp_oed
```

### 3. Conclusion
Italy's 13.3% NEET rate is the mathematical result of having both a rigid O$\rightarrow$E vector (class-based tracking) and a massive O$\rightarrow$D vector (nepotism/shadow economy), effectively breaking the E$\rightarrow$D meritocratic link.
