### 1. Youth Cohabitation & Generational Wealth Transfer
We model the percentage of young adults (18-34) living with their parents across Europe, indicating a total reliance on Origin wealth to survive the NEET/unemployment period.

```python
cohabitation = {
    'Entity': ['Italy', 'Spain', 'EU-27', 'Germany', 'Sweden'],
    'Living_With_Parents_18_34_Pct': [67.4, 64.5, 49.0, 41.2, 17.5], # ISTAT/Eurostat proxies
    'State_Welfare_Efficacy_Rank': ['Low', 'Low', 'Medium', 'High', 'Very High']
}
df_cohabitation = pd.DataFrame(cohabitation)
df_cohabitation.to_csv('local_data/processed/blind_spots/youth_cohabitation.csv', index=False)
df_cohabitation
```

### 2. Analysis
The data proves that the ultimate safety net for a failed Destination (D) is simply falling back to Origin (O). With 67.4% of young Italians forced to live with their parents, the system relies on the pension/savings of the older generation to subsidize the unemployment of the younger. 

If a student's Origin lacks this generational wealth (e.g., lower-class families in the South), the OED penalty becomes absolute poverty.
