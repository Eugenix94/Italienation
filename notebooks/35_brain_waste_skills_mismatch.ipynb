# 35. Brain Waste & Skills Mismatch (Blind Spot)
This notebook analyzes the phenomenon of "Brain Waste" (Over-education). It mathematically models what happens when a student *does* succeed in the Education vector (achieving a tertiary degree) but is absorbed by an Italian labor market that does not produce high-knowledge jobs.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/blind_spots', exist_ok=True)
```

### 1. The Skills Mismatch: Over-Education in the Labor Market
We model the percentage of tertiary-educated youth (25-34) who are employed in jobs that do not require a degree (e.g., baristas, retail, low-level administration).

```python
brain_waste = {
    'Entity': ['Italy', 'Spain', 'EU-27', 'Germany', 'Netherlands'],
    'Tertiary_Overeducation_Rate_Pct': [35.1, 36.2, 22.4, 16.5, 15.2], # Percent of graduates in low-skilled jobs
    'Knowledge_Economy_Index_Rank': [28, 26, 12, 6, 4] # Propensity of the economy to generate high-value jobs
}
df_brain_waste = pd.DataFrame(brain_waste)
df_brain_waste.to_csv('local_data/processed/blind_spots/tertiary_overeducation.csv', index=False)
df_brain_waste
```

### 2. Analysis
The Italian Destination (D) is mathematically capped by the nature of the economy itself. With over 35% of graduates working in jobs that don't require their degree, the E $\rightarrow$ D meritocratic promise is broken. This "Brain Waste" is the direct domestic twin of the "Brain Drain" (emigration); both are symptoms of a labor market fundamentally unable to absorb high human capital.
