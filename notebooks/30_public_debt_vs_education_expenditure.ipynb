### 2. Analysis
The data presents a devastating mathematical reality: Italy spends more on simply servicing the *interest* of its past debt (4.3% of GDP) than it does on funding the entire education system (4.1% of GDP). 

This structural starvation is the root cause of the broken OED pipeline. The state literally cannot afford to modernize infrastructure or pay for vocational training (as Germany does), forcing it to institutionalize unpaid PCTO labor.
