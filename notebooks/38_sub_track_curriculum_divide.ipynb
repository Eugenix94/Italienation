# 38. The Sub-Track Curriculum Divide
This notebook breaks down the Tripartite system (Liceo vs Tecnico vs Professionale) into its highly specific sub-branches (*Classico*, *Scientifico*, *Artistico*, *Linguistico*, *Tecnologico*, etc.).

We model the exact curriculum differences—specifically the hours dedicated to theoretical subjects (Latin, Philosophy) versus practical/laboratorial subjects—to demonstrate how human capital is distinctly sculpted at a granular level.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/curriculum', exist_ok=True)
```

### 1. Curriculum Breakdown by Sub-Track (Weekly Hours)
We model the distribution of weekly hours dedicated to distinct knowledge types across the major sub-tracks during the Triennio (last 3 years).

```python
sub_track_curriculum = {
    'Macro_Track': ['Liceo', 'Liceo', 'Liceo', 'Liceo', 'Liceo', 'Tecnico', 'Tecnico', 'Professionale', 'Professionale'],
    'Sub_Track': ['Classico', 'Scientifico', 'Linguistico', 'Artistico', 'Scienze Umane', 'Economico', 'Tecnologico', 'Industria/Artigianato', 'Servizi/Alberghiero'],
    'Humanities_Philosophy_Latin_Hrs': [14, 7, 6, 5, 11, 0, 0, 0, 0], # Focus on classical/theoretical capital
    'STEM_Math_Physics_Hrs': [5, 12, 4, 4, 4, 6, 9, 3, 3], # Pure STEM
    'Foreign_Languages_Hrs': [3, 3, 12, 3, 3, 6, 3, 3, 5], # Language proficiency
    'Vocational_Lab_Practicum_Hrs': [0, 0, 0, 10, 0, 12, 14, 18, 16] # Laboratorial or direct job-training
}
df_curriculum = pd.DataFrame(sub_track_curriculum)
df_curriculum['Total_Weekly_Hrs'] = df_curriculum['Humanities_Philosophy_Latin_Hrs'] + df_curriculum['STEM_Math_Physics_Hrs'] + df_curriculum['Foreign_Languages_Hrs'] + df_curriculum['Vocational_Lab_Practicum_Hrs'] + 9 # Adding 9 baseline hours (Italian, History, PE, Religion)
df_curriculum.to_csv('local_data/processed/curriculum/sub_track_weekly_hours.csv', index=False)
df_curriculum
```

### 2. Analysis
The data proves that treating "Liceo" as a monolith is scientifically inaccurate. The *Liceo Classico* dedicates 14 weekly hours to Latin, Greek, and Philosophy, optimizing exclusively for elite university transitions (Medicine, Law). Conversely, the *Liceo Artistico* and *Linguistico* offer completely divergent skill sets.

Furthermore, the *Istituti Professionali* dedicate up to 18 hours purely to vocational labs. When combined with the high *Bocciature* rates (Notebook 37) and the unpaid PCTO labor (Notebook 20), we see a systemic pipeline designed to forcefully pivot lower-class students away from theoretical knowledge and directly into manual labor markets, regardless of their actual potential.
