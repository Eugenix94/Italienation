# 18. Esplorazione Geospaziale e Case Study Catania
### Mappatura Territoriale della Segregazione Culturale e del Modello Tripartito

Questo quaderno interattivo utilizza i dati geolocalizzati del Ministero dell'Istruzione (`Scuola in Chiaro` via HuggingFace) per visualizzare l'impronta territoriale del modello scolastico italiano.

L'analisi geospaziale è cruciale per comprendere come il capitale umano (`Licei` vs `Istituti Tecnici` vs `Istituti Professionali`) sia distribuito non solo a livello astratto, ma fisicamente nei tessuti urbani e periferici.

---
### Struttura dell'Analisi:
1. **Mappa Nazionale Generalizzata**: Un campione statistico (~15%) delle scuole secondarie superiori italiane, sufficiente per mostrare la densità del sistema educativo senza sovraccaricare il browser o esporre eccessivamente le coordinate esatte.
2. **Catania Case Study**: Un focus approfondito su tutte le 1.694 scuole (infanzia, primaria, secondaria) e nello specifico sugli Istituti Superiori della provincia di Catania, per osservare la stratificazione sociale urbana e la distribuzione del capitale culturale teorico (`Licei`) rispetto a quello operativo (`Professionali / LeFP`).

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
import matplotlib.pyplot as plt
import os

base_url = "https://raw.githubusercontent.com/Eugenix94/Italienation/main/local_data/processed/"
# Fallback per l'esecuzione locale nel repository
if os.path.exists("../local_data/processed/catania_geospatial_schools_case_study.csv"):
    base_url = "../local_data/processed/"

print("Librerie geospaziali caricate (Pandas, Folium, MarkerCluster).")
```

## 1. Catania Case Study: Distribuzione Fisica del Capitale Umano
Carichiamo il dataset filtrato per la Provincia di Catania. Analizziamo come gli istituti superiori si posizionano sul territorio. Spesso, i Licei (capitale culturale simbolico) si concentrano nei centri storici o nei quartieri agiati, mentre gli Istituti Professionali e Tecnici sono più diffusi nelle cinture periferiche.

```python
catania_df = pd.read_csv(base_url + "catania_geospatial_schools_case_study.csv")

# Filtriamo solo le scuole secondarie superiori per la mappa
superiori = ['Liceo (Accademico)', 'Istituto Tecnico', 'Istituto Professionale']
ct_sup = catania_df[catania_df['TRACK_TRIPARTITO'].isin(superiori)].dropna(subset=['LATITUDINE', 'LONGITUDINE'])

print(f"Istituti Superiori analizzati nella Provincia di Catania: {len(ct_sup)}")

# Configurazione colori per Track
color_map = {
    'Liceo (Accademico)': 'blue',
    'Istituto Tecnico': 'green',
    'Istituto Professionale': 'red'
}

# Creiamo la mappa centrata su Catania
map_ct = folium.Map(location=[37.507877, 15.083030], zoom_start=11, tiles="CartoDB positron")

for idx, row in ct_sup.iterrows():
    track = row['TRACK_TRIPARTITO']
    # Se la colonna nome scuola non c'è, usiamo il tipo
    nome = row.get('DENOMINAZIONESCUOLA', track) 
    
    folium.CircleMarker(
        location=[row['LATITUDINE'], row['LONGITUDINE']],
        radius=6,
        popup=f"<b>{nome}</b><br>Track: {track}",
        color=color_map.get(track, 'gray'),
        fill=True,
        fill_color=color_map.get(track, 'gray'),
        fill_opacity=0.7
    ).add_to(map_ct)

map_ct
```

> **Analisi Sociologica (Catania)**: La mappa cromatica rivela spesso "cluster" di capitale umano. Si può osservare empiricamente come la separazione precoce a 14 anni (`Tripartizione`) non sia solo una scelta astratta, ma si traduca in una segregazione geografica quotidiana tra gli studenti dei diversi indirizzi, limitando la coesione sociale trasversale che invece il *Comprehensive System* britannico garantisce raccogliendo gli studenti nello stesso edificio.

## 2. Mappa Nazionale (Campione Generalizzato)
Visualizziamo l'estensione dell'apparato scolastico superiore italiano utilizzando un campione statistico geolocalizzato.
La densità dei punti (`MarkerCluster`) mostra la straordinaria capillarità dell'istruzione tecnica e professionale nella Penisola.

```python
nat_sample = pd.read_csv(base_url + "italy_national_schools_geospatial_sample.csv")
nat_sample = nat_sample.dropna(subset=['LATITUDINE', 'LONGITUDINE'])

map_it = folium.Map(location=[41.8719, 12.5674], zoom_start=6, tiles="CartoDB dark_matter")
marker_cluster = MarkerCluster().add_to(map_it)

for idx, row in nat_sample.iterrows():
    track = row['TRACK_TRIPARTITO']
    folium.CircleMarker(
        location=[row['LATITUDINE'], row['LONGITUDINE']],
        radius=4,
        color=color_map.get(track, 'white'),
        fill=True,
        fill_opacity=0.6
    ).add_to(marker_cluster)

map_it
```
