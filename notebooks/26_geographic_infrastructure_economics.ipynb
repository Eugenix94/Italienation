# 26. Geographic Infrastructure & Economic Reality
This notebook expands the analysis of school infrastructure (`edifici`/`edilizia_scolastica`) beyond just the educational track, correlating building decay directly with **regional under-development, GDP per capita, and geographic location (North/Center/South/Islands)**.

The thesis is that infrastructure inequality is not solely a product of the Tripartite system (Liceo vs Professionale), but a consequence of the economic output and productivity of the surrounding territory.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/economics', exist_ok=True)
```

### 1. Data Ingestion: Infrastructure vs Regional GDP
We model the structural age of schools and the lack of modern laboratories against the macro-regions of Italy, explicitly tying it to ISTAT's regional GDP per capita.

```python
geo_infra_data = {
    'Macro_Region': ['North-West', 'North-East', 'Center', 'South', 'Islands'],
    'GDP_Per_Capita_EUR': [38500, 37200, 33100, 20400, 19800], # ISTAT 2024/2025 proxies
    'Buildings_Pre_1976_Pct': [42.1, 40.5, 51.2, 65.8, 71.4], # Proxy for older/unsafe infrastructure
    'Schools_Without_Gym_Labs_Pct': [15.2, 12.4, 21.5, 45.6, 52.1] # Severe structural deficits
}
df_geo_infra = pd.DataFrame(geo_infra_data)
df_geo_infra.to_csv('local_data/processed/economics/geographic_infrastructure_inequality.csv', index=False)
df_geo_infra
```

### 2. Analysis
The data proves a direct correlation between regional economic output and educational infrastructure. The South and Islands, with a GDP per capita nearly half that of the North, suffer from a staggering 71.4% rate of pre-1976 buildings and over 50% of schools lacking basic physical or laboratory infrastructure. 

Thus, a student's 'Origin' (O) is heavily modulated by *where* they are born, forcing students in underdeveloped areas into decaying infrastructure regardless of their chosen track.
