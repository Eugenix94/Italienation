# 17. Frammentazione Curricolare e Sviluppo del Capitale Umano
### Confronto Strutturale tra il Comprehensive System Britannico (Fino a 16 Anni) e il Sistema Tripartito Italiano (Dal 14 Anni)

Questo quaderno analizza il divario strutturale nella distribuzione delle materie e delle ore di insegnamento tra l'ordinamento scolastico inglese (`National Curriculum KS3/KS4 + GCSEs`) e quello italiano (`Liceo Classico/Scientifico, Istituto Tecnico, Istituto Professionale`).

---
### Il Problema Sociologico e Pedagocico:
1. **Regno Unito (UK — Comprehensive System)**: Fino all'età di 16 anni, *tutti gli studenti* frequentano la stessa tipologia di scuola unica (`Comprehensive School`) e seguono un **National Curriculum comune e poliedrico** che comprende 11+ materie: Inglese, Matematica, Scienze (Biologia, Chimica, Fisica), Storia, Geografia, Lingue Straniere, Informatica (`Computing`), Arte, Musica e Tecnologia & Progettazione (`Design & Technology`). Questo garantisce una **base universale di capitale culturale e scientifico** prima di ogni scelta di specializzazione (`A-Levels` a 16 anni).
2. **Italia (IT — Sistema Tripartito Precoce a 14 Anni)**: La separazione irreversibile a 14 anni crea una **marcata segregazione curricolare**:
   - I **Licei** concentrano le ore su materie classico-retoriche (fino a 18 ore tra Latino, Greco, Filosofia, Letteratura) ma marginalizzano le discipline STEM e azzerano l'economia, il diritto e la tecnologia pratica.
   - Gli **Istituti Tecnici e Professionali** forniscono competenze operative e tecnologiche di settore, ma escludono totalmente il capitale culturale teorico-filosofico (`nessuna ora di Filosofia o Storia dell'Arte`), limitando gli strumenti critici per la mobilità sociale ascendente (`Habitus di Bourdieu`).

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

base_url = "https://raw.githubusercontent.com/Eugenix94/Italienation/main/local_data/processed/"
print("Setup ambiente completato per Notebook 17.")
```

## 1. Il Quadro Orario e Disciplinare a Confronto (Fascia 14-16 e Superiore)

```
df_curr = pd.read_csv(base_url + "curriculum_subjects_tripartite_vs_comprehensive_panel.csv")
display(df_curr[['Paese', 'Sistema_o_Indirizzo', 'Età_Studenti', 'Totale_Materie_Distinte_Insegnate', 'Indice_Poliedricita_Curricolare_0_10', 'Segregazione_Capitale_Culturale_Score_0_10']])
```

## 2. Ripartizione del Tempo Scuola (Ore Settimanali per Area del Sapere)
Il seguente grafico a barre impilate (`Stacked Bar Chart`) mostra come il Comprehensive System britannico mantenga una distribuzione equilibrata tra Scienze/Informatica, Umanistica, Arte/Progettazione ed Economia/Diritto, mentre il sistema italiano polarizzi e segreghi il sapere a seconda del canale scelto a 14 anni.

```
cols_ore = [
    'Ore_Umanistiche_Classiche',
    'Ore_STEM_e_Informatica',
    'Ore_Economia_Diritto',
    'Ore_Arte_Tecnologia_Progettazione',
    'Ore_Civica_e_Laboratori_Pratici'
]

df_plot = df_curr.set_index('Sistema_o_Indirizzo')[cols_ore]

colors = ['#818cf8', '#34d399', '#fbbf24', '#f87171', '#a78bfa']
ax = df_plot.plot(kind='bar', stacked=True, figsize=(14, 7), color=colors, edgecolor='black', linewidth=0.5)

plt.title("Ripartizione Oraria Settimanale per Aree Disciplinari: Comprehensive vs Canali Italiani", fontsize=15, fontweight="bold")
plt.ylabel("Ore Settimanali di Insegnamento")
plt.xlabel("Ordinamento e Indirizzo Scolastico")
plt.xticks(rotation=20, ha='right')
plt.legend([
    'Umanistiche & Classiche (Latino/Greco/Filosofia/Storia)',
    'STEM & Informatica (Matematica/Fisica/Scienze/Coding)',
    'Economia & Diritto (Business/Civic Economy)',
    'Arte, Tecnologia & Progettazione (Design & Technology/Musica)',
    'Civica, Laboratori Operativi & Sport'
], bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 3. Indice di Poliedricità Curricolare vs Segregazione del Capitale Culturale
Mettiamo a confronto diretto l'Indice di Poliedricità (`0-10`, capacità di fornire una base completa scientifico-umanistica) con il Punteggio di Segregazione del Capitale Culturale (`0-10`, tendenza a privare un canale sociale di strumenti teorici o tecnici chiave).

```
df_indices = df_curr[['Sistema_o_Indirizzo', 'Indice_Poliedricita_Curricolare_0_10', 'Segregazione_Capitale_Culturale_Score_0_10']].melt(id_vars='Sistema_o_Indirizzo')

plt.figure(figsize=(14, 6))
sns.barplot(data=df_indices, x='Sistema_o_Indirizzo', y='value', hue='variable', palette=['#38bdf8', '#f43f5e'])
plt.title("Confronto tra Poliedricità Curricolare (Blu) e Segregazione del Capitale Culturale (Rosso)", fontsize=14, fontweight="bold")
plt.ylabel("Punteggio Indice (0-10)")
plt.xlabel("Ordinamento e Indirizzo")
plt.xticks(rotation=20, ha='right')
plt.legend(['Indice di Poliedricità Curricolare (Più è alto, più il curricolo è completo)', 'Indice di Segregazione del Capitale Culturale (Più è alto, più frammenta la società)'])
plt.tight_layout()
plt.show()
```

## 4. Considerazioni Sociologiche: Il Circolo Vizioso di Bourdieu e il Futuro del Capitale Umano
I dati empirici confermano che il modello britannico post-1988 (`Education Reform Act`), obbligando **tutti gli studenti ad affrontare un curricolo STEM + Umanistico + Design & Technology (D&T) uniforme fino ai 16 anni**, riduce la riproduzione passiva di disuguaglianze di classe legate alla scelta scolastica precoce.

In Italia, viceversa, il doppio binario precoce a 14 anni produce una **doppia deprivazione di capitale umano**:
1. **Per le classi popolari negli Istituti Professionali/Tecnici**: la totale assenza di Filosofia e Storia dell'Arte preclude l'acquisizione del linguaggio critico-retorico e della flessibilità epistemologica, relegandoli a mansioni esecutive e vulnerabili all'automazione IA.
2. **Per le élite nei Licei Classici/Scientifici**: la quasi totale assenza di progettazione tecnologica pratica (`Design & Technology`), informatica moderna ed economia civile genera una leadership colta ma sprovvista degli strumenti algoritmici e finanziari necessari a guidare l'innovazione industriale del Paese.

---
### Sintesi Politico-Educativa per il Manifesto:
La transizione da un **sistema tripartito frammentato (14 anni)** verso un **biennio unico o modello Comprehensive fino a 16 anni** rappresenta la leva fondamentale per sbloccare la stagnazione della produttività (`TFP`) e democratizzare il capitale umano in Italia.
