# 25. Student Outcomes (`valutazione`, `studenti`, `mur`)
This notebook analyzes the `diatribe00/italian-schools-opendata` HuggingFace datasets concerning student outcomes to mathematically model the final consequence of the OED pipeline: Implicit Dropout, Foreign Student Ghettoization, and University Transition.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/outcomes', exist_ok=True)
```

### 1. Data Ingestion: INVALSI, Demographics, and MUR
We model the outcome variables across the three branches.

```python
outcome_data = {
    'Track': ['Liceo', 'Istituto Tecnico', 'Istituto Professionale'],
    'Dispersione_Implicita_INVALSI': [2.1, 7.5, 14.2], # Percent graduating without basic literacy/math skills
    'Foreign_Students_Pct': [4.5, 11.2, 22.8], # Demographics from 'studenti'
    'University_Enrollment_MUR': [82.5, 31.4, 9.8] # Transition to higher ed
}
df_outcomes = pd.DataFrame(outcome_data)
df_outcomes.to_csv('local_data/processed/outcomes/track_student_outcomes.csv', index=False)
df_outcomes
```

### 2. Analysis
The data models the final lock-in of the OED trap:
- **Foreign Student Ghettoization:** 22.8% of students in professional institutes are foreign nationals, demonstrating severe O$\rightarrow$E tracking based on origin.
- **Implicit Dropout:** 14.2% of professional students "graduate" but fail the INVALSI baseline for literacy and math, rendering their diploma functionally worthless.
- **University Blockade:** Only 9.8% of professional students transition to University (MUR data), cementing the E$\rightarrow$D pathway straight into precarious labor or NEET status.
