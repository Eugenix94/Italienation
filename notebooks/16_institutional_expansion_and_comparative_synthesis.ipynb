# 16. Sintesi dei Nuovi Dati Istituzionali e Normativi: Le 5 Frontiere
### L'Atlante in Continua Espansione — Analisi Econometrica e Ordinamentale Comparata

Questo notebook esplora in dettaglio le **5 Nuove Frontiere Strategiche** acquisite tramite API istituzionali (`SDMX 2.1 ISTAT`, `SPARQL/REST MIM SNAES`, `MUR/INDIRE`, `Infratel/AGCOM`, `legislation.gov.uk API`).

---
### Obiettivi Analitici:
1. **Edilizia Scolastica & PNRR**: Valutazione del divario territoriale di agibilità antisismica e impatto degli stanziamenti PNRR Mense/Palestre.
2. **Inverno Demografico**: Modellizzazione econometrica della contrazione scolastica 2025–2040 e impatto degli accorpamenti.
3. **Terziario Tecnico (ITS Academy vs Università)**: Analisi comparata dei tassi occupazionali a 1 anno e differenziali salariali.
4. **Divario Digitale & Banda Ultra-Larga 1 Gbps**: Correlazione tra infrastruttura di rete scolastica (FTTH) e competenze DigComp.
5. **Timeline Giuridico-Storica Comparata**: Confronto strutturale degli ordinamenti scolastici (UK Education Acts vs Italia D.Lgs 297/94 e Riforma Gentile).

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

base_url = "https://raw.githubusercontent.com/Eugenix94/Italienation/main/local_data/processed/"
print("Setup ambiente completato per Notebook 16.")
```

## 1. Edilizia Scolastica e Sicurezza Antisismica (MIM SNAES & PNRR Missione 4)
I dati ufficiali dell'Anagrafe Nazionale dell'Edilizia Scolastica (`SNAES`) del Ministero dell'Istruzione rivelano un marcato gradiente Nord-Sud nella certificazione antisismica e di agibilità degli edifici scolastici.

```
df_infra = pd.read_csv(base_url + "school_infrastructure_seismic_safety_panel.csv")
display(df_infra)

plt.figure(figsize=(12, 6))
sns.barplot(data=df_infra, x="Regione", y="Perc_Certificato_Agibilita", palette="viridis")
plt.title("Quota di Edifici Scolastici con Certificato di Agibilità per Regione (%)", fontsize=14, fontweight="bold")
plt.ylabel("Percentuale (%)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 2. Inverno Demografico e Accorpamenti Scolastici 2025–2040 (ISTAT SDMX API)
Le proiezioni demografiche ufficiali ISTAT stimano una contrazione del -30.5% della popolazione scolastica in Calabria e del -27.0% in Sicilia entro il 2035, comportando la chiusura o l'accorpamento di centinaia di Istituti Comprensivi e aumentando il tempo di pendolarismo per gli studenti delle aree interne.

```
df_demo = pd.read_csv(base_url + "demographic_winter_school_closures_projection.csv")
display(df_demo)

plt.figure(figsize=(12, 6))
sns.barplot(data=df_demo, x="Regione", y="Variazione_Percentuale_2024_2035", palette="magma")
plt.title("Proiezione Variazione Popolazione Scolastica (6-18 anni) 2024–2035 (%)", fontsize=14, fontweight="bold")
plt.ylabel("Variazione (%)")
plt.axhline(0, color="black", linestyle="--")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 3. Esiti Occupazionali e Stipendi: ITS Academy vs Università Triennali (INDIRE / MUR)
Il confronto tra il terziario tecnologico professionale (`ITS Academy`) e i corsi di laurea triennale generalisti mostra un forte differenziale occupazionale e reddituale a 1 anno dal diploma, evidenziando il divario di iscritti rispetto al sistema duale europeo (`Duale Ausbildung` DE).

```
df_its = pd.read_csv(base_url + "its_academy_vs_university_outcomes.csv")
display(df_its)

plt.figure(figsize=(12, 6))
sns.barplot(data=df_its, x="Tasso_Occupazione_1_Anno_Perc", y="Settore_Terziario", palette="crest")
plt.title("Tasso di Occupazione a 1 Anno dal Diploma/Laurea (%)", fontsize=14, fontweight="bold")
plt.xlabel("Tasso Occupazione (%)")
plt.tight_layout()
plt.show()
```

## 4. Divario Digitale e Banda Ultra-Larga FTTH 1 Gbps nelle Scuole (Infratel / AGCOM)
L'implementazione del Piano Scuola Connessa di Infratel Italia mostra come la copertura in fibra ottica 1 Gbps nelle aule scolastiche sia direttamente correlata all'indice di resilienza e alfabetizzazione digitale degli studenti (`DigComp`).

```
df_dig = pd.read_csv(base_url + "digital_divide_broadband_schools_nuts3.csv")
display(df_dig)

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_dig, x="Scuole_Connesse_1Gbps_Perc", y="Indice_DigComp_Studenti", hue="Regione", s=150, palette="tab10")
plt.title("Correlazione tra Connessione Scuole 1 Gbps (%) e Indice DigComp Studenti", fontsize=14, fontweight="bold")
plt.xlabel("Scuole Connesse a 1 Gbps (%)")
plt.ylabel("Indice DigComp Medio")
plt.tight_layout()
plt.show()
```
