# 20. PCTO & Human Capital: The Differential of Unpaid Labor
This notebook analyzes the discrepancy in human capital development across the Italian Tripartite system, explicitly focusing on the **Percorsi per le Competenze Trasversali e per l'Orientamento (PCTO)** mandated by L. 107/2015 (Buona Scuola).

We mathematically contrast the Italian unpaid model against international work-based learning frameworks (e.g., German Dual System, Eurydice Framework).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

os.makedirs('local_data/processed/pcto', exist_ok=True)
```

### 1. The Statutory PCTO Hours Differential
According to the Italian legal framework, the minimum mandatory hours of PCTO (unpaid work-related learning) over the final 3 years of upper secondary school are strictly stratified by track.

```python
pcto_data = {
    'Track': ['Liceo', 'Istituto Tecnico', 'Istituto Professionale'],
    'Mandatory_PCTO_Hours': [90, 150, 210],
    'Compensation': ['Unpaid', 'Unpaid', 'Unpaid']
}
df_pcto = pd.DataFrame(pcto_data)

plt.figure(figsize=(10, 6))
sns.barplot(data=df_pcto, x='Track', y='Mandatory_PCTO_Hours', palette=['#34d399', '#fbbf24', '#f87171'])
plt.title('Mandatory Unpaid PCTO Hours by Track (L. 107/2015)', fontsize=14, fontweight='bold')
plt.ylabel('Minimum Hours (Last 3 Years)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.text(-0.2, 100, 'Academic Focus', color='white', fontweight='bold')
plt.text(0.8, 160, 'Mixed Focus', color='white', fontweight='bold')
plt.text(1.8, 220, 'Labor Focus', color='white', fontweight='bold')
plt.savefig('local_data/processed/pcto/pcto_hours_differential.png', dpi=300, bbox_inches='tight')
plt.show()
```

### 2. The Legal Framework: Institutionalized Unpaid Labor
The mandatory nature of PCTO is established by two key laws:
1. **L. 107/2015 (La Buona Scuola):** Originally mandated 200 hours for Licei and 400 hours for Tecnici/Professionali of *Alternanza Scuola-Lavoro*.
2. **Legge di Bilancio 2019 (L. 145/2018):** Reduced the hours to 90h, 150h, and 210h respectively, and rebranded it as *PCTO*. Crucially, the law maintains that these activities are **mandatory for admission to the State Exam**, yet provides **no legal requirement for remuneration**, distinguishing it completely from employment contracts.

### 3. Eurydice Comparative Framework: The European Context
According to the **Eurydice Network** and **Cedefop (European Centre for the Development of Vocational Training)**, work-based learning (WBL) in Europe is highly diverse, but heavily trends toward paid apprenticeships in robust systems:

- **Paid Apprenticeship Models (Dual Systems):** Germany, Austria, Switzerland, and Denmark mandate strict remuneration. In Germany, the Federal Vocational Training Act (BBiG) sets a minimum allowance (€682/month in 2025, €724 in 2026).
- **School-Based Unpaid Models:** Italy's PCTO falls into a unique category of mandatory, unpaid school-based placements for *all* tracks, not just vocational ones. While other countries have unpaid internships (like France's *stages*), Italy's strict requirement of 210 hours specifically for the working-class track (Professionale) without compensation is an anomaly compared to the EFQEA (European Framework for Quality and Effective Apprenticeships) push for paid conditions.

### 4. Synthesis: The Human Capital Penalty
By legally mandating (L. 145/2018) 210 hours of unretributed labor for students in the *Professionale* track, the Italian system explicitly subtracts 210 hours of potential academic/theoretical study (Cultural Capital). This establishes a structural "penalty" for working-class tracks, conditioning students to accept precarious, unpaid labor ("paid in experience") at age 15, directly preceding their entry into an unregulated labor market with high NEET risks.
