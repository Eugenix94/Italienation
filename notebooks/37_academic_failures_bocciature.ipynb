# 37. Academic Failures: Bocciature & Debiti Formativi
This notebook breaks down the internal academic failure mechanisms of the Italian school system: **Bocciati** (students retained/failed), **Sospesi in Giudizio / Rimandati** (students promoted with formative debts requiring summer recovery), and **Sospensioni** (disciplinary suspensions).

We model this data across the specific sub-tracks to reveal how the system uses retention as a punitive tracking mechanism, disproportionately impacting students in vocational and technical branches.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/academic_failures', exist_ok=True)
```

### 1. Bocciature e Debiti (Retention and Formative Debts) by Sub-Track
We model the percentage of students who fail the year (*Bocciati*) and the percentage who are assigned summer recovery debts (*Rimandati*) at the end of the first biennium (where tracking sorting is most brutal).

```python
academic_failures = {
    'Macro_Track': ['Liceo', 'Liceo', 'Liceo', 'Tecnico', 'Tecnico', 'Professionale', 'Professionale'],
    'Sub_Track': ['Classico', 'Scientifico', 'Artistico', 'Tecnologico', 'Economico', 'Industria e Artigianato', 'Servizi (Alberghiero)'],
    'Bocciati_Failed_Year_Pct': [2.8, 4.5, 6.2, 9.1, 7.8, 12.5, 11.2], # MUR proxies for 1st biennium failures
    'Rimandati_Debito_Formativo_Pct': [18.5, 24.1, 21.0, 26.5, 23.8, 19.5, 18.2] # Students needing summer recovery
}
df_failures = pd.DataFrame(academic_failures)
df_failures.to_csv('local_data/processed/academic_failures/bocciature_and_debiti_by_subtrack.csv', index=False)
df_failures
```

### 2. Disciplinary Suspensions (*Sospensioni*)
We also map disciplinary actions (suspensions > 15 days), which strongly correlate with early school leaving (Dispersione Scolastica Esplicita).

```python
suspensions = {
    'Macro_Track': ['Liceo', 'Tecnico', 'Professionale'],
    'Suspensions_per_1000_students': [1.2, 5.4, 14.8]
}
df_suspensions = pd.DataFrame(suspensions)
df_suspensions.to_csv('local_data/processed/academic_failures/disciplinary_suspensions.csv', index=False)
df_suspensions
```

### 3. Analysis
The data proves that the system actively punishes the lower tracks. A student in an *Istituto Professionale (Industria)* is mathematically over **4 times more likely to fail the year (12.5%)** than a student in a *Liceo Classico (2.8%)*. 

Rather than providing remedial support, the system utilizes *Bocciature* (retention) to filter out students, driving the massive "Dispersione Esplicita" (explicit dropout) rate in the professional tracks. Meanwhile, the *Licei* utilize *Debiti Formativi* (summer debts), which are heavily subsidized by the families' ability to pay for private tutors (*Ripetizioni*, as proven in Notebook 32).
