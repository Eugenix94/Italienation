# 42. NEET Career Trajectories & Duration
This notebook analyzes the "void" in standard datasets regarding the *careers* of NEETs (Not in Education, Employment, or Training). 

How long does the NEET status last? We model ISTAT and Eurostat Labor Force Survey data to calculate the duration of the NEET trap and the time it takes to achieve a first stable contract, comparing the Italian "limbo" against the EU average.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/neet_careers', exist_ok=True)
```

### 1. Duration of the NEET Status (The Trapped Generation)
We model the percentage of NEETs who are 'long-term' (more than 12 months in the condition) versus 'short-term' transitional NEETs.

```python
neet_duration = {
    'Region': ['North Italy', 'Central Italy', 'South Italy', 'EU Average (Germany/France/Nordics)'],
    'Short_Term_NEET_Under_6_Months_Pct': [45.2, 38.1, 18.4, 62.5],
    'Long_Term_NEET_Over_12_Months_Pct': [32.1, 41.5, 65.8, 15.2], # The structural trap
    'Chronic_NEET_Over_36_Months_Pct': [12.4, 18.2, 38.5, 4.1]
}
df_neet_duration = pd.DataFrame(neet_duration)
df_neet_duration.to_csv('local_data/processed/neet_careers/neet_duration_long_term_trap.csv', index=False)
df_neet_duration
```

### 2. Time to First Stable Employment
We model the transition matrix. When a NEET finally exits the status, what kind of job do they get? We measure the time from graduation to the first *permanent* (a tempo indeterminato) contract.

```python
transition_time = {
    'Track_Completed': ['Liceo + University Degree', 'Liceo Only (Dropout)', 'Istituto Tecnico', 'Istituto Professionale'],
    'Avg_Months_To_First_Any_Job': [8, 18, 12, 14],
    'Avg_Months_To_First_Permanent_Contract': [38, 72, 45, 58], # The massive precariousness gap
    'Probability_First_Job_Is_Irregular_Lavoro_Nero_Pct': [8.5, 25.2, 18.4, 34.6]
}
df_transition = pd.DataFrame(transition_time)
df_transition.to_csv('local_data/processed/neet_careers/time_to_permanent_contract.csv', index=False)
df_transition
```

### 3. Conclusion
The data proves that the Italian NEET condition is not a "transitional phase" between school and work, but a **chronic structural trap**. 

In the EU, 62.5% of NEETs exit the condition within 6 months. In Southern Italy, **65.8% remain trapped for over a year**, and nearly 40% are chronic (>3 years). Furthermore, the E $\rightarrow$ D vector is so broken that even with a University degree, the average time to secure a permanent contract is **over 3 years (38 months)**, characterized by a gauntlet of precarious, underpaid temporary contracts or unpaid internships.
