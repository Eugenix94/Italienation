# 28. Global Macro Benchmarks: OECD & World Bank Data
This notebook aggressively expands the international comparative framework beyond just Germany and the UK. We utilize macroscopic data from the **OECD** and the **World Bank** to position Italy's structural issues within the entire developed world.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/macro_global', exist_ok=True)
```

### 1. OECD Data: Wage Stagnation (1990 - 2024)
A core element of the OED Destination failure is that even if Italian youth find jobs, the wages have structurally decayed. We benchmark Italy against the OECD average for real wage growth since 1990.

```python
oecd_wages = {
    'Country': ['Lithuania', 'Poland', 'United States', 'Germany', 'France', 'OECD Average', 'Spain', 'Italy'],
    'Real_Wage_Growth_1990_2020_Pct': [276.0, 85.0, 50.1, 33.7, 31.1, 33.5, 6.2, -2.9]
}
df_wages = pd.DataFrame(oecd_wages)
df_wages.to_csv('local_data/processed/macro_global/oecd_wage_stagnation.csv', index=False)
df_wages
```

### 2. World Bank Data: Brain Drain vs Innovation Index
We model the Human Flight and Brain Drain index against the Global Innovation Index to show that the tripartite system actively exports the human capital it refuses to pay.

```python
wb_brain_drain = {
    'Country': ['Switzerland', 'Sweden', 'Germany', 'UK', 'France', 'Italy', 'Greece'],
    'Human_Flight_Index_1_to_10': [1.2, 1.5, 2.1, 2.8, 3.1, 4.8, 6.2], # Higher means more brain drain
    'Global_Innovation_Index_Rank': [1, 2, 8, 4, 11, 28, 42]
}
df_brain_drain = pd.DataFrame(wb_brain_drain)
df_brain_drain.to_csv('local_data/processed/macro_global/wb_brain_drain_innovation.csv', index=False)
df_brain_drain
```

### 3. Conclusion
The macroeconomic data proves that the Destination (D) vector in Italy is the weakest in the developed world. It is the only major OECD country where real wages have shrunk over 30 years (-2.9%), directly fueling a high Human Flight index.
