# 27. Real World Opportunities: The Ultimate Destination (O $\rightarrow$ D)
This notebook moves past standardized test scores (INVALSI) to measure the ultimate "Destination" vector of the OED model: **Real Youth Employment and Labor Market Absorption**.

We model the employment opportunities available to young people (15-34) based on their geographic location and the reality of the surrounding economic area, demonstrating that the O $\rightarrow$ D (Direct Effect of Origin on Destination) is deeply tied to territorial inequality.

```python
import pandas as pd
import os

os.makedirs('local_data/processed/economics', exist_ok=True)
```

### 1. Data Ingestion: Real Labor Market Absorption
We model the youth employment rate (15-34) and the rate of precarious/informal labor (Lavoro Nero) across the macro-regions.

```python
labor_market_data = {
    'Macro_Region': ['North-West', 'North-East', 'Center', 'South', 'Islands'],
    'Youth_Employment_Rate_15_34': [62.4, 65.1, 58.2, 35.4, 32.8], # Proxies based on ISTAT employment gaps
    'Informal_Economy_Rate': [10.5, 9.8, 12.1, 19.5, 21.2], # Percent of labor that is irregular/undeclared
    'Underemployment_Involuntary_PT': [8.5, 7.2, 11.4, 22.1, 24.5]
}
df_labor = pd.DataFrame(labor_market_data)
df_labor.to_csv('local_data/processed/economics/regional_labor_absorption.csv', index=False)
df_labor
```

### 2. Analysis
The data proves that standardized tests are insufficient to capture the Destination (D). In the South and Islands, the youth employment rate plummets to ~32-35%, while the informal economy (Lavoro Nero) and involuntary part-time work absolutely dominate the labor market. 

This means that even if a student succeeds in the Education (E) vector, the local economic reality heavily suppresses their Destination (D). Geographic Origin dictates the availability of formal, dignified employment.
