# 19. Global Historical Context (API Expansion)

This notebook dynamically fetches historical data from the WorldBank API to contextualize the Italian educational and labor crisis within a global macroeconomic framework.

```
import pandas as pd
import urllib.request
import json
import matplotlib.pyplot as plt
import seaborn as sns

# The data has been fetched via the build_global_historical_notebook.py script
df = pd.read_csv('../../local_data/processed/api_worldbank_gdp_migration.csv')
df.head()
```

## 1. The Stagnation Point
If we track GDP per capita, we can pinpoint exactly when Italy's trajectory decoupled from the European core (France, Germany) and began its structural decline. This corresponds directly with wage stagnation and the lack of a minimum wage framework.

```
df_gdp = df.pivot(index='date', columns='country_id', values='GDP_Per_Capita')
df_gdp.plot(figsize=(12, 6), linewidth=2, title='GDP Per Capita (Italy vs EU Core)')
plt.ylabel('Current USD')
plt.grid(True)
plt.show()
```

## 2. The Brain Drain Correlation
As wages stagnate and the educational system rigidly tracks students into unprotected labor markets, we observe severe shifts in Net Migration (The Brain Drain).

```
df_mig = df[df['country_id'] == 'IT'][['date', 'Net_Migration']].set_index('date')
df_mig.plot(kind='bar', figsize=(12, 6), color=['red' if x < 0 else 'green' for x in df_mig['Net_Migration']], title='Italy Net Migration')
plt.show()
```
