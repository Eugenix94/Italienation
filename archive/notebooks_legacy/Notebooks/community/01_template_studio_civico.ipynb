```python
# 1. Importazione delle librerie necessarie (Standard Data Science / Pandas & Matplotlib)
import pandas as pd
import matplotlib.pyplot as plt

# Configurazione grafica chiara per i grafici
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12
print("✅ Librerie importate con successo e pronte per l'analisi civica!")
```

## 📂 Passo 1: Carica i Dati Ufficiali (`Load Canonical Data`)
Qui sotto carichiamo direttamente dal repository i dati ufficiali sui bilanci scolastici comunali (`siope_school_expenditure_summary.csv`), che contengono la spesa reale di **tutti i 7.959 Comuni italiani**.

```python
# Caricamento diretto dall'URL immutabile di Italienation (Conforme a OSF/Zenodo)
url_siope = "https://raw.githubusercontent.com/Eugenix94/Italienation/main/holistic_analysis/data_panels/siope_school_expenditure_summary.csv"
df_comuni = pd.read_csv(url_siope)

print(f"📊 Dati comunali caricati! Numero totale di Comuni nel dataset: {len(df_comuni)}")
# Visualizziamo le prime 5 righe del dataset
df_comuni.head()
```

## 🔍 Passo 2: Cerca ed Esplora il Tuo Comune o la Tua Regione (`Explore Your Area`)
In questa cella puoi cambiare il nome della **Regione** (`es. 'Campania', 'Lombardia', 'Sicilia'`) o del **Comune** per vedere subito quanto si spende per studente nella tua zona per mense, manutenzione e scuolabus!

```python
# 👉 MODIFICA QUI: Scrivi il nome della tua Regione o lascia vuoto per vedere tutta Italia
mia_regione = "Campania"  # Esempio: "Lombardia", "Sicilia", "Lazio", "Puglia"

if "Regione" in df_comuni.columns:
    df_filtrato = df_comuni[df_comuni["Regione"].str.contains(mia_regione, case=False, na=False)]
    print(f"📌 Comuni trovati nella Regione '{mia_regione}': {len(df_filtrato)}")
    # Mostriamo i 10 Comuni con la spesa scolastica pro-capite o totale più alta nella Regione selezionata
    if "Spesa_Totale_Scuola_EUR" in df_filtrato.columns:
        display(df_filtrato.sort_values(by="Spesa_Totale_Scuola_EUR", ascending=False).head(10))
    else:
        display(df_filtrato.head(10))
else:
    print("Colonne disponibili:", df_comuni.columns)
    display(df_comuni.head(10))
```

## 📈 Passo 3: Crea un Grafico Chiaro e Spiega i Risultati (`Create Your Visual Chart`)
Adesso creiamo un grafico di confronto tra le diverse Regioni o Città per vedere le diseguaglianze con i nostri occhi e spiegarle ai cittadini e alle istituzioni.

```python
# Confronto della spesa media o distribuzione tra Regioni (se presente colonna Regione)
if "Regione" in df_comuni.columns and "Spesa_Totale_Scuola_EUR" in df_comuni.columns:
    spesa_regionale = df_comuni.groupby("Regione")["Spesa_Totale_Scuola_EUR"].mean().sort_values(ascending=False).head(12)
    
    plt.figure(figsize=(12, 6))
    spesa_regionale.plot(kind='bar', color='#3b82f6', edgecolor='black')
    plt.title("Spesa Media Comunale per la Scuola (Top 12 Regioni)", fontsize=14, fontweight='bold')
    plt.ylabel("Spesa Media (Euro)", fontsize=12)
    plt.xlabel("Regione", fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
else:
    print("💡 Grafico di riepilogo generale per le prime 10 righe:")
    df_comuni.select_dtypes(include=['number']).iloc[:10, :3].plot(kind='bar', figsize=(10, 5))
    plt.title("Riepilogo Indicatori Numerici nei Primi 10 Comuni")
    plt.show()
```
