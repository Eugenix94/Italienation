## Data Scope

Lower secondary source (ISTAT):
- flow `52_1044_DF_DCIS_SCUOLE_10` (`Secondaria I grado - indicatori scolastici`)
- outcome proxy used: `failure_at_exam_proxy = 100 - licenziati per 100 esaminati`

Upper secondary source (ISTAT):
- flow `52_1044_DF_DCIS_SCUOLE_15` (`Secondaria II grado - ripetenti per anno di corso`)
- outcome signal used: repeater rate (`ripetenti`)

Common reference:
- `TIME_PERIOD = 2024` interpreted as proxy year `2024/2025`.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', context='talk')
pd.set_option('display.max_columns', 60)

ROOT = Path.cwd()
if ROOT.name == 'Notebooks':
    ROOT = ROOT.parent

UPPER_LONG = ROOT / 'local_data' / 'ISTAT' / 'school_outcomes' / 'istat_repeaters_upper_secondary_long.csv'
LOWER_LONG = ROOT / 'local_data' / 'ISTAT' / 'school_outcomes' / 'istat_lower_secondary_indicators_long.csv'
LOWER_LATEST = ROOT / 'local_data' / 'processed' / 'istat_lower_secondary_indicators_latest.csv'
LOWER_EXAM_PROXY = ROOT / 'local_data' / 'processed' / 'istat_lower_secondary_exam_proxy_latest.csv'

upper = pd.read_csv(UPPER_LONG)
lower_long = pd.read_csv(LOWER_LONG)
lower_latest = pd.read_csv(LOWER_LATEST)
lower_exam_proxy = pd.read_csv(LOWER_EXAM_PROXY)

upper['TIME_PERIOD'] = pd.to_numeric(upper['TIME_PERIOD'], errors='coerce')
upper['OBS_VALUE'] = pd.to_numeric(upper['OBS_VALUE'], errors='coerce')
lower_long['TIME_PERIOD'] = pd.to_numeric(lower_long['TIME_PERIOD'], errors='coerce')
lower_long['OBS_VALUE'] = pd.to_numeric(lower_long['OBS_VALUE'], errors='coerce')
lower_latest['TIME_PERIOD'] = pd.to_numeric(lower_latest['TIME_PERIOD'], errors='coerce')
lower_latest['OBS_VALUE'] = pd.to_numeric(lower_latest['OBS_VALUE'], errors='coerce')
lower_latest['failure_at_exam_proxy'] = pd.to_numeric(lower_latest['failure_at_exam_proxy'], errors='coerce')
lower_exam_proxy['TIME_PERIOD'] = pd.to_numeric(lower_exam_proxy['TIME_PERIOD'], errors='coerce')
lower_exam_proxy['failure_at_exam_proxy'] = pd.to_numeric(lower_exam_proxy['failure_at_exam_proxy'], errors='coerce')

latest_year = int(max(upper['TIME_PERIOD'].max(), lower_latest['TIME_PERIOD'].max()))
proxy_label = f'{latest_year}/{latest_year + 1}'

OUT_DIR = ROOT / 'Notebooks' / 'transition_outputs'
OUT_DIR.mkdir(parents=True, exist_ok=True)

def save_chart(name: str):
    path = OUT_DIR / name
    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches='tight')
    print(f'Saved {path}')

print('Latest proxy year:', proxy_label)
print('Output dir:', OUT_DIR)
```

```python
# National comparables
lower_nat = lower_latest[(lower_latest['REF_AREA'] == 'IT') & (lower_latest['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')]
lower_nat_exam = float(lower_nat.loc[lower_nat['DATA_TYPE'] == 'EXAM', 'OBS_VALUE'].iloc[0])
lower_nat_failure = 100 - lower_nat_exam

upper_nat = upper[(upper['REF_AREA'] == 'IT') & (upper['TIME_PERIOD'] == latest_year) & (upper['SEX'] == 'T') & (upper['CITIZENSHIP'] == 'TOTAL') & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL') & (upper['SCHOOL_YEAR'] == 'ALL')]
upper_nat_all = float(upper_nat.loc[upper_nat['TYPE_SCHOOL'] == 'ALL', 'OBS_VALUE'].iloc[0])
upper_nat_lic = float(upper_nat.loc[upper_nat['TYPE_SCHOOL'] == 'LIC', 'OBS_VALUE'].iloc[0])
upper_nat_tec = float(upper_nat.loc[upper_nat['TYPE_SCHOOL'] == 'TEC', 'OBS_VALUE'].iloc[0])
upper_nat_voc = float(upper_nat.loc[upper_nat['TYPE_SCHOOL'] == 'VOC', 'OBS_VALUE'].iloc[0])

national_compare = pd.DataFrame([
    {'metric': 'Lower secondary exam failure proxy', 'value': lower_nat_failure},
    {'metric': 'Upper secondary repeaters (all)', 'value': upper_nat_all},
    {'metric': 'Upper secondary repeaters (licei)', 'value': upper_nat_lic},
    {'metric': 'Upper secondary repeaters (tecnici)', 'value': upper_nat_tec},
    {'metric': 'Upper secondary repeaters (professionali)', 'value': upper_nat_voc},
])
national_compare
```

```python
fig, ax = plt.subplots(figsize=(11, 6))
sns.barplot(data=national_compare, y='metric', x='value', hue='metric', legend=False, palette='Set2', ax=ax)
ax.set_title(f'National contrast: middle-school failure proxy vs upper-secondary repeaters ({proxy_label})', fontweight='bold')
ax.set_xlabel('Rate / value')
ax.set_ylabel('')
save_chart('transition_01_national_contrast.png')
plt.show()
```

## Regional Transition Stress Index

We define a simple transition index as:
- `transition_stress_index = upper_secondary_repeaters_all - lower_secondary_failure_proxy`

Large values indicate territories where lower-secondary failure remains very low but upper-secondary repetition pressure is much higher.

```python
lower_reg = lower_latest[(lower_latest['DATA_TYPE'] == 'EXAM') & (lower_latest['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')]
lower_reg = lower_reg[lower_reg['REF_AREA'].str.len() == 4][['REF_AREA', 'REF_AREA_LABEL', 'OBS_VALUE']].copy()
lower_reg['lower_exam_failure_proxy'] = 100 - lower_reg['OBS_VALUE']

upper_reg = upper[(upper['TIME_PERIOD'] == latest_year) & (upper['SEX'] == 'T') & (upper['CITIZENSHIP'] == 'TOTAL') & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL') & (upper['SCHOOL_YEAR'] == 'ALL') & (upper['TYPE_SCHOOL'] == 'ALL')]
upper_reg = upper_reg[upper_reg['REF_AREA'].str.len() == 4][['REF_AREA', 'REF_AREA_LABEL', 'OBS_VALUE']].copy()
upper_reg = upper_reg.rename(columns={'OBS_VALUE': 'upper_repeaters_all'})

transition = upper_reg.merge(lower_reg[['REF_AREA', 'lower_exam_failure_proxy']], on='REF_AREA', how='inner')
transition['transition_stress_index'] = transition['upper_repeaters_all'] - transition['lower_exam_failure_proxy']
transition = transition.sort_values('transition_stress_index', ascending=False)
transition.head(10)
```

```python
fig, ax = plt.subplots(figsize=(12, 7))
sns.barplot(data=transition, y='REF_AREA_LABEL', x='transition_stress_index', hue='REF_AREA_LABEL', legend=False, palette='magma', ax=ax)
ax.set_title(f'Regional transition stress index ({proxy_label})', fontweight='bold')
ax.set_xlabel('Upper repeaters (ALL) - lower exam failure proxy')
ax.set_ylabel('Region')
save_chart('transition_02_regional_stress_index.png')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 7))
sns.scatterplot(data=transition, x='lower_exam_failure_proxy', y='upper_repeaters_all', hue='transition_stress_index', palette='viridis', s=120, ax=ax)
for row in transition.itertuples():
    ax.text(row.lower_exam_failure_proxy + 0.005, row.upper_repeaters_all + 0.03, row.REF_AREA_LABEL, fontsize=8)
ax.set_title(f'Regional mapping: lower-school failure proxy vs upper repeaters ({proxy_label})', fontweight='bold')
ax.set_xlabel('Lower-secondary exam failure proxy')
ax.set_ylabel('Upper-secondary repeaters (ALL)')
save_chart('transition_03_scatter_lower_vs_upper.png')
plt.show()
```

## Direct Transition Link: Lagged Bridge (t-1 -> t)

To make the lower/upper linkage more concrete, we connect:
- lower-secondary exam failure proxy at year `t-1`
- first-year upper-secondary repeaters (`SCHOOL_YEAR = FIR`) at year `t`

This better matches the transition timing from middle school to first year of upper secondary.

```python
# Lower proxy at t-1 by region (historical long series)
lower_yearly = lower_long[
    (lower_long['DATA_TYPE'] == 'EXAM')
    & (lower_long['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (lower_long['REF_AREA'].str.len() == 4)
][['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']].copy()
lower_yearly['lower_exam_failure_proxy'] = 100 - lower_yearly['OBS_VALUE']

# First-year repeaters at t by region
upper_fir = upper[
    (upper['SCHOOL_YEAR'] == 'FIR')
    & (upper['TYPE_SCHOOL'] == 'ALL')
    & (upper['SEX'] == 'T')
    & (upper['CITIZENSHIP'] == 'TOTAL')
    & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (upper['REF_AREA'].str.len() == 4)
][['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']].copy()
upper_fir = upper_fir.rename(columns={'OBS_VALUE': 'upper_fir_repeaters'})

# Merge t (upper) with t-1 (lower)
lower_lag = lower_yearly.copy()
lower_lag['TIME_PERIOD'] = lower_lag['TIME_PERIOD'] + 1

bridge = upper_fir.merge(
    lower_lag[['REF_AREA', 'TIME_PERIOD', 'lower_exam_failure_proxy']],
    on=['REF_AREA', 'TIME_PERIOD'],
    how='inner'
)
bridge['transition_jump_fir'] = bridge['upper_fir_repeaters'] - bridge['lower_exam_failure_proxy']

lag_corr = bridge[['lower_exam_failure_proxy', 'upper_fir_repeaters']].corr().iloc[0, 1]
print(f'Lagged correlation (lower t-1 vs upper FIR t): {lag_corr:.3f}')

bridge_latest = bridge[bridge['TIME_PERIOD'] == latest_year].copy()
bridge_latest = bridge_latest.sort_values('transition_jump_fir', ascending=False)
bridge_latest.head(10)
```

```python
fig, ax = plt.subplots(figsize=(10, 7))
sns.scatterplot(
    data=bridge_latest,
    x='lower_exam_failure_proxy',
    y='upper_fir_repeaters',
    hue='transition_jump_fir',
    palette='viridis',
    s=120,
    ax=ax
)
for row in bridge_latest.itertuples():
    ax.text(row.lower_exam_failure_proxy + 0.005, row.upper_fir_repeaters + 0.03, row.REF_AREA_LABEL, fontsize=8)

ax.set_title(
    f'Lagged transition link: lower failure (t-1) vs upper first-year repeaters (t) - {proxy_label}',
    fontweight='bold'
)
ax.set_xlabel('Lower-secondary exam failure proxy (t-1)')
ax.set_ylabel('Upper-secondary first-year repeaters (t)')
save_chart('transition_06_lagged_link_fir.png')
plt.show()
```

## Model-Ready Panel Export (Add All)

This section exports a compact region-year panel for modeling the lower -> upper transition.

Target file:
- `local_data/processed/transition_bridge_model_panel.csv`

Panel design:
- predictors from lower secondary at `t-1`
- outcomes from upper secondary at `t`
- regional key = `REF_AREA` (NUTS-like ISTAT territorial code)

```python
# Build lower-secondary predictors by year/region (ALL management)
lower_for_panel = lower_long[
    (lower_long['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (lower_long['REF_AREA'].str.len() == 4)
    & (lower_long['DATA_TYPE'].isin(['EXAM', 'FOR', 'ENROL', 'PUB', 'DISAB', 'LMG']))
][['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'DATA_TYPE', 'OBS_VALUE']].copy()

lower_panel = lower_for_panel.pivot_table(
    index=['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD'],
    columns='DATA_TYPE',
    values='OBS_VALUE',
    aggfunc='mean'
).reset_index()

lower_panel = lower_panel.rename(columns={
    'EXAM': 'lower_exam_success_t_minus_1',
    'FOR': 'lower_foreign_share_t_minus_1',
    'ENROL': 'lower_class_size_t_minus_1',
    'PUB': 'lower_public_share_t_minus_1',
    'DISAB': 'lower_disability_per_1000_t_minus_1',
    'LMG': 'lower_median_grade_t_minus_1'
})
lower_panel['lower_exam_failure_t_minus_1'] = 100 - lower_panel['lower_exam_success_t_minus_1']

# Shift lower predictors to match upper outcomes at t
lower_panel['TIME_PERIOD'] = lower_panel['TIME_PERIOD'] + 1

# Upper outcomes at t
upper_all_t = upper[
    (upper['SCHOOL_YEAR'] == 'ALL')
    & (upper['TYPE_SCHOOL'] == 'ALL')
    & (upper['SEX'] == 'T')
    & (upper['CITIZENSHIP'] == 'TOTAL')
    & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (upper['REF_AREA'].str.len() == 4)
][['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']].copy().rename(columns={'OBS_VALUE': 'upper_repeaters_all_t'})

upper_fir_t = upper[
    (upper['SCHOOL_YEAR'] == 'FIR')
    & (upper['TYPE_SCHOOL'] == 'ALL')
    & (upper['SEX'] == 'T')
    & (upper['CITIZENSHIP'] == 'TOTAL')
    & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (upper['REF_AREA'].str.len() == 4)
][['REF_AREA', 'TIME_PERIOD', 'OBS_VALUE']].copy().rename(columns={'OBS_VALUE': 'upper_repeaters_fir_t'})

upper_tracks_t = upper[
    (upper['SCHOOL_YEAR'] == 'ALL')
    & (upper['TYPE_SCHOOL'].isin(['LIC', 'TEC', 'VOC']))
    & (upper['SEX'] == 'T')
    & (upper['CITIZENSHIP'] == 'TOTAL')
    & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL')
    & (upper['REF_AREA'].str.len() == 4)
][['REF_AREA', 'TIME_PERIOD', 'TYPE_SCHOOL', 'OBS_VALUE']].copy()

upper_tracks_wide = upper_tracks_t.pivot_table(
    index=['REF_AREA', 'TIME_PERIOD'],
    columns='TYPE_SCHOOL',
    values='OBS_VALUE',
    aggfunc='mean'
).reset_index()
upper_tracks_wide = upper_tracks_wide.rename(columns={'LIC': 'upper_lic_t', 'TEC': 'upper_tec_t', 'VOC': 'upper_voc_t'})
upper_tracks_wide['upper_voc_minus_lic_t'] = upper_tracks_wide['upper_voc_t'] - upper_tracks_wide['upper_lic_t']
upper_tracks_wide['upper_tec_minus_lic_t'] = upper_tracks_wide['upper_tec_t'] - upper_tracks_wide['upper_lic_t']

# Final panel merge
model_panel = lower_panel.merge(
    upper_all_t,
    on=['REF_AREA', 'REF_AREA_LABEL', 'TIME_PERIOD'],
    how='inner'
).merge(
    upper_fir_t,
    on=['REF_AREA', 'TIME_PERIOD'],
    how='left'
).merge(
    upper_tracks_wide,
    on=['REF_AREA', 'TIME_PERIOD'],
    how='left'
)

model_panel = model_panel.sort_values(['TIME_PERIOD', 'REF_AREA']).reset_index(drop=True)

# Useful derived columns for quick diagnostics
model_panel['transition_jump_all_t'] = model_panel['upper_repeaters_all_t'] - model_panel['lower_exam_failure_t_minus_1']
model_panel['transition_jump_fir_t'] = model_panel['upper_repeaters_fir_t'] - model_panel['lower_exam_failure_t_minus_1']

panel_path = ROOT / 'local_data' / 'processed' / 'transition_bridge_model_panel.csv'
model_panel.to_csv(panel_path, index=False)
print('Saved model panel:', panel_path)
print('Shape:', model_panel.shape)

latest_panel = model_panel[model_panel['TIME_PERIOD'] == latest_year].copy()
latest_top = latest_panel.sort_values('transition_jump_fir_t', ascending=False).head(15)

latest_top_path = ROOT / 'local_data' / 'processed' / 'transition_bridge_latest_top_jump.csv'
latest_top.to_csv(latest_top_path, index=False)
print('Saved latest top-jump table:', latest_top_path)

model_panel.head(10)
```

This panel keeps the upper-secondary track split visible so transition stress can be interpreted against licei/tecnici/professionali divergence.

```python
upper_types_reg = upper[(upper['TIME_PERIOD'] == latest_year) & (upper['SEX'] == 'T') & (upper['CITIZENSHIP'] == 'TOTAL') & (upper['TYPE_SCHOOL_MANAGEMENT'] == 'ALL') & (upper['SCHOOL_YEAR'] == 'ALL') & (upper['TYPE_SCHOOL'].isin(['LIC','TEC','VOC']))]
upper_types_reg = upper_types_reg[upper_types_reg['REF_AREA'].str.len() == 4].copy()
type_map = {'LIC': 'Licei', 'TEC': 'Tecnici', 'VOC': 'Professionali'}
upper_types_reg['track'] = upper_types_reg['TYPE_SCHOOL'].map(type_map)

fig, ax = plt.subplots(figsize=(13, 7))
sns.barplot(data=upper_types_reg, x='REF_AREA_LABEL', y='OBS_VALUE', hue='track', ax=ax)
ax.set_title(f'Upper-secondary repeaters by track and region ({proxy_label})', fontweight='bold')
ax.set_xlabel('Region')
ax.set_ylabel('Repeaters rate')
ax.tick_params(axis='x', rotation=65)
ax.legend(title='Track')
save_chart('transition_04_upper_track_by_region.png')
plt.show()
```

## ANPAL Replacement: Youth Engagement & Labour-Market Absorption

ANPAL Garanzia Giovani data is not available from public sources. Instead, we use official substitutes:
- **NEET rates** (Eurostat/ISTAT) = youth unavailability proxy
- **Youth unemployment** (Eurostat) = job-placement inverse
- **Early school leavers** (Eurostat) = pre-program risk indicator
- **NEET by migration** (Eurostat) = program equity proxy

These datasets capture the same youth population and outcomes ANPAL would measure.

```python
import json
from pathlib import Path

# Load ANPAL replacement datasets
processed_dir = Path("../local_data/processed")

# 1. Load annual NEET rate
anpal_neet = pd.read_csv(processed_dir / "anpal_replacement_neet_annual.csv")
print(f"NEET annual series: {len(anpal_neet)} years")
print(anpal_neet)

# 2. Load NEET by migration (equity breakdown)
try:
    anpal_neet_mig = pd.read_csv(processed_dir / "anpal_replacement_neet_by_migration.csv")
    print(f"\nNEET by migration: {len(anpal_neet_mig)} rows")
    display(anpal_neet_mig.head(10))
except Exception as e:
    print(f"⚠ NEET by migration: {e}")

# 3. Load youth unemployment (placement proxy)
try:
    anpal_yunemp = pd.read_csv(processed_dir / "anpal_replacement_youth_unemployment.csv")
    print(f"\nYouth unemployment: {len(anpal_yunemp)} rows")
    display(anpal_yunemp.head(10))
except Exception as e:
    print(f"⚠ Youth unemployment: {e}")

# 4. Load early school leavers (risk indicator)
try:
    anpal_esl = pd.read_csv(processed_dir / "anpal_replacement_early_school_leavers.csv")
    print(f"\nEarly school leavers: {len(anpal_esl)} rows")
    display(anpal_esl.head(10))
except Exception as e:
    print(f"⚠ Early school leavers: {e}")

# 5. Load replacement dataset manifest (explanation)
with open(processed_dir / "anpal_replacement_manifest.json") as f:
    manifest = json.load(f)
print(f"\n📋 Manifest: {manifest['title']}")
print(f"   Context: {manifest['context']}")
```
