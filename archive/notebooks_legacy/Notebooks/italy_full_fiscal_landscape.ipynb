# Italy Education Fiscal Landscape

This notebook consolidates the current fiscal evidence on Italy's education system into one reproducible analytical surface. It combines macro public and private spending, school-ledger proxies, level-aware institutional finance, household cost burden, tertiary support, and territorial funding-risk proxies.

## Source reliability rule
- Main fiscal analysis uses locally processed OECD, World Bank, Eurostat, SIOPE, MUR, MIM, and DSU outputs.
- Save the Children indicators are split by validation tier. Only indicators reproducible from local INVALSI or ISTAT files are treated as validated. PDF-only claims remain quarantined and are not mixed into the main fiscal evidence.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='deep')
pd.set_option('display.max_columns', 100)
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

ROOT = Path.cwd().parents[0] if Path.cwd().name == 'Notebooks' else Path.cwd()
LOCAL = ROOT / 'local_data'
PROC = LOCAL / 'processed'
INVALSI = LOCAL / 'INVALSI'
EUROSTAT = LOCAL / 'eurostat'
ISTAT = LOCAL / 'ISTAT'

def pct_to_float(series: pd.Series) -> pd.Series:
    return pd.to_numeric(series.astype(str).str.replace(',', '.', regex=False).str.strip(), errors='coerce')

def bn(series: pd.Series) -> pd.Series:
    return series / 1_000_000_000

print('ROOT:', ROOT)
print('Processed outputs:', PROC.exists())
```

```python
inventory = pd.read_csv(PROC / 'education_fiscal_inventory.csv')
history = pd.read_csv(PROC / 'italy_education_expenditure_history_panel.csv')
finance = pd.read_csv(PROC / 'italy_education_finance_levels_real.csv')
household = pd.read_csv(PROC / 'italy_household_burden_module.csv')
territorial = pd.read_csv(PROC / 'italy_education_territorial_proxy_panel.csv')
save_children_pdf = pd.read_csv(PROC / 'save_the_children_italy_school_stats.csv')

invalsi_trend = pd.read_csv(INVALSI / 'dati-sottostanti-le-dashboard-di-tableau-del-rapporto-2023-2024-grado-8-grado-13__trend_traguardi_ms2024-dati-pop-g08-e-g13-2.csv', sep=';')
for col in ['Perc.2018', 'Perc.2019', 'Perc.2021', 'Perc.2022', 'Perc.2023', 'Perc.2024']:
    invalsi_trend[col] = pct_to_float(invalsi_trend[col])

istat_esl = pd.read_csv(ISTAT / 'istat_early_school_leavers.csv')

print('Inventory rows:', len(inventory))
print('History rows:', len(history))
print('Finance rows:', len(finance))
print('Household rows:', len(household))
print('Territorial rows:', len(territorial))
print('Save the Children PDF rows:', len(save_children_pdf))
```

## 1. System-Wide Fiscal Burden

This section focuses on the macro history: total public education spending as a share of GDP, public spending as a share of total government expenditure, and the OECD split between state funding and household or private funding.

```python
macro = history.sort_values('year').copy()
latest_macro = macro.loc[macro['year'] == macro['year'].max()]
display(macro.tail(10))

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].plot(macro['year'], macro['public_pct_gdp_owid'], label='Public spending (% GDP, OWID/WB)', linewidth=2)
axes[0].plot(macro['year'], macro['state_pct_gdp'], label='State funding (% GDP, OECD)', linewidth=2)
axes[0].plot(macro['year'], macro['parents_private_pct_gdp'], label='Parents/private (% GDP, OECD)', linewidth=2)
axes[0].plot(macro['year'], macro['total_pct_gdp'], label='Total (% GDP, OECD)', linestyle='--', linewidth=2)
axes[0].set_title('Italy education spending burden over time')
axes[0].set_ylabel('% of GDP')
axes[0].legend(loc='best')

axes[1].plot(macro['year'], macro['public_pct_govexp_owid'], color='#b24d2c', linewidth=2)
axes[1].set_title('Public education spending as share of government expenditure')
axes[1].set_ylabel('% of government expenditure')

for ax in axes:
    ax.grid(alpha=0.25)
    ax.set_xlabel('Year')

plt.tight_layout()
plt.show()

macro_summary = pd.DataFrame({
    'metric': [
        'Latest public spending % GDP (OWID/WB)',
        'Latest public spending % govt expenditure (OWID/WB)',
        'Latest state funding % GDP (OECD)',
        'Latest parents/private funding % GDP (OECD)',
        'Latest total funding % GDP (OECD)'
    ],
    'value': [
        macro.loc[macro['public_pct_gdp_owid'].notna(), 'public_pct_gdp_owid'].iloc[-1],
        macro.loc[macro['public_pct_govexp_owid'].notna(), 'public_pct_govexp_owid'].iloc[-1],
        macro.loc[macro['state_pct_gdp'].notna(), 'state_pct_gdp'].iloc[-1],
        macro.loc[macro['parents_private_pct_gdp'].notna(), 'parents_private_pct_gdp'].iloc[-1],
        macro.loc[macro['total_pct_gdp'].notna(), 'total_pct_gdp'].iloc[-1],
    ]
})
display(macro_summary)
```

## 3. Level-Aware Institutional Finance In Real Terms

The new level-aware panel separates Eurostat institutional expenditure by sector and education level, then deflates the values into constant 2022 euros using a World Bank CPI series. The notebook keeps SIOPE school-ledger values and tuition proxies separate because they describe different fiscal surfaces.

```python
latest_eurostat_year = int(finance.loc[finance['series_group'] == 'eurostat_level_history', 'year'].max())
latest_levels = finance[(finance['series_group'] == 'eurostat_level_history') & (finance['year'] == latest_eurostat_year) & (finance['sector'].isin(['PUBL', 'PRIV', 'TOT_SEC']))].copy()
latest_levels['real_bn_eur_2022'] = bn(latest_levels['real_base_year_eur'])
pivot_levels = latest_levels.pivot_table(index='level_label', columns='sector', values='real_bn_eur_2022', aggfunc='sum').reset_index()
display(pivot_levels.sort_values('TOT_SEC', ascending=False).head(15))

plot_levels = latest_levels.groupby(['level_label', 'sector'], as_index=False)['real_bn_eur_2022'].sum()
plot_levels = plot_levels[plot_levels['level_label'].isin(plot_levels.groupby('level_label')['real_bn_eur_2022'].sum().sort_values(ascending=False).head(8).index)]

plt.figure(figsize=(14, 6))
sns.barplot(data=plot_levels, x='real_bn_eur_2022', y='level_label', hue='sector')
plt.title(f'Eurostat institutional expenditure by level and sector, Italy {latest_eurostat_year} (real 2022 EUR bn)')
plt.xlabel('Real expenditure (EUR bn, 2022 prices)')
plt.ylabel('Education level')
plt.tight_layout()
plt.show()

siope_proxy = finance[finance['series_group'] == 'siope_school_proxy'][['year', 'nominal_eur', 'real_base_year_eur', 'per_student_eur', 'note']].copy()
display(siope_proxy.tail(5))
```

## 4. Household Burden And Tertiary Support

This module combines the macro private-share view with concrete family-facing cost proxies: primary textbook cover prices, estimated secondary annual spending, average university tuition, and regional DSU or ERSU support.

```python
display(household)

micro_costs = household[household['domain'] == 'household_micro_cost'].copy()
micro_costs['mid_eur'] = micro_costs[['min_eur', 'max_eur']].mean(axis=1)
micro_costs = micro_costs[micro_costs['mid_eur'].notna()]

plt.figure(figsize=(13, 6))
sns.barplot(data=micro_costs.sort_values('mid_eur', ascending=False), x='mid_eur', y='segment', hue='indicator', dodge=False)
plt.title('Household-facing education cost proxies')
plt.xlabel('EUR')
plt.ylabel('Segment')
plt.tight_layout()
plt.show()

support_rows = household[household['domain'].isin(['tertiary_access_buffer', 'tertiary_support'])][['domain', 'indicator', 'value', 'unit', 'note']].copy()
display(support_rows)
```

## 5. Territorial Funding-Risk Proxy

The territorial panel joins regional school stocks and outcome indicators to named regional tertiary support spending. It does not claim to be a full regional school-expenditure panel. The funding side is explicitly a DSU or ERSU proxy because the available SIOPE regional pivots are not yet reliable enough for a direct named join.

```python
territorial_rank = territorial[['region_name', 'neet_count_15_29_thousands', 'neet_risk_index', 'school_students_total', 'estimated_bes_share_pct', 'estimated_disabled_share_pct', 'upper_secondary_repeaters_pct', 'dsu_total_spend_eur', 'dsu_spend_per_eligible_eur']].copy()
territorial_rank['dsu_total_spend_eur_bn'] = bn(territorial_rank['dsu_total_spend_eur'])
display(territorial_rank.sort_values('dsu_total_spend_eur', ascending=False).head(10))

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
top_support = territorial_rank.sort_values('dsu_total_spend_eur', ascending=False).head(10)
sns.barplot(data=top_support, x='dsu_total_spend_eur_bn', y='region_name', ax=axes[0], color='#4C78A8')
axes[0].set_title('Top regions by observed DSU/ERSU support spending')
axes[0].set_xlabel('EUR bn')
axes[0].set_ylabel('Region')

sns.scatterplot(data=territorial_rank, x='dsu_spend_per_eligible_eur', y='neet_risk_index', size='school_students_total', hue='upper_secondary_repeaters_pct', ax=axes[1], palette='rocket')
axes[1].set_title('Funding proxy vs NEET risk')
axes[1].set_xlabel('DSU spend per eligible student (EUR)')
axes[1].set_ylabel('NEET risk index')

plt.tight_layout()
plt.show()
```

## 6. Save The Children Validation Layer

The earlier workspace file used a PDF-only extraction. This section rebuilds the usable subset from local official data:
- **INVALSI** validates the lower-secondary Italian and math under-threshold shares.
- **ISTAT** supplies the official local early-school-leaver rate, but the local file currently stops at 2020 rather than the PDF's 2021 label.
- **Archipelago learning-gain claims** remain PDF-only program-evaluation statements and are kept outside the validated evidence tier.

```python
invalsi_it = invalsi_trend[(invalsi_trend['GRADO'] == 'Terza Secondaria di Primo Grado') & (invalsi_trend['Ripartizione'] == 'Italia') & (invalsi_trend['MATERIA'].isin(['Italiano', 'Matematica']))].copy()

invalsi_it['below_minimum_2022'] = 100 - invalsi_it['Perc.2022']

invalsi_it['below_minimum_2024'] = 100 - invalsi_it['Perc.2024']

display(invalsi_it[['MATERIA', 'Perc.2022', 'below_minimum_2022', 'Perc.2024', 'below_minimum_2024']])



for col in ['ITTER107', 'CITTADINANZA', 'CONDIZIONE_PROF', 'SESSO', 'TIPO_DATO', 'TIME_PERIOD', 'OBS_VALUE']:

    istat_esl[col] = pd.to_numeric(istat_esl[col], errors='coerce') if col != 'ITTER107' else istat_esl[col].astype(str)



istat_rate = istat_esl[

    (istat_esl['ITTER107'] == 'IT')

    & (istat_esl['CITTADINANZA'] == 9)

    & (istat_esl['CONDIZIONE_PROF'] == 99)

    & (istat_esl['SESSO'] == 9)

    & (istat_esl['TIPO_DATO'] == 29)

].copy()

istat_rate = istat_rate.sort_values('TIME_PERIOD')

display(istat_rate[['TIME_PERIOD', 'OBS_VALUE']].tail(10))



pdf_lookup = save_children_pdf.set_index('indicator')['value'].to_dict()

below_it_2022 = float(invalsi_it.loc[invalsi_it['MATERIA'] == 'Italiano', 'below_minimum_2022'].iloc[0])

below_math_2022 = float(invalsi_it.loc[invalsi_it['MATERIA'] == 'Matematica', 'below_minimum_2022'].iloc[0])



if istat_rate.empty:

    esl_latest_local_year = np.nan

    esl_latest_local_value = np.nan

    esl_note = 'No local ISTAT early-school-leaver rate rows matched the expected filter in this checkout.'

else:

    esl_latest_local_year = int(istat_rate['TIME_PERIOD'].max())

    esl_latest_local_value = float(istat_rate.loc[istat_rate['TIME_PERIOD'] == esl_latest_local_year, 'OBS_VALUE'].iloc[0])

    esl_note = 'Local ISTAT official rate is available through 2020 in this checkout; it is close in level but does not directly reproduce the PDF\'s 2021 timestamp.'



validation_rows = [

    {

        'indicator': 'students_below_minimum_italian_lower_secondary_pct',

        'pdf_value': pdf_lookup.get('students_below_minimum_italian_lower_secondary_pct'),

        'local_validated_value': below_it_2022,

        'local_reference_year': 2022,

        'validation_status': 'validated_from_local_invalsi',

        'validation_note': 'Computed as 100 minus INVALSI share reaching the target in Italian for grade 8, Italy, 2022.'

    },

    {

        'indicator': 'students_below_minimum_math_lower_secondary_pct',

        'pdf_value': pdf_lookup.get('students_below_minimum_math_lower_secondary_pct'),

        'local_validated_value': below_math_2022,

        'local_reference_year': 2022,

        'validation_status': 'validated_from_local_invalsi',

        'validation_note': 'Computed as 100 minus INVALSI share reaching the target in mathematics for grade 8, Italy, 2022.'

    },

    {

        'indicator': 'early_school_leavers_18_24_pct',

        'pdf_value': pdf_lookup.get('early_school_leavers_18_24_pct'),

        'local_validated_value': esl_latest_local_value,

        'local_reference_year': esl_latest_local_year,

        'validation_status': 'partially_validated_local_official_latest_available' if not np.isnan(esl_latest_local_value) else 'not_validated_local_missing',

        'validation_note': esl_note

    },

    {

        'indicator': 'archipelago_learning_gain_math_months',

        'pdf_value': pdf_lookup.get('archipelago_learning_gain_math_months'),

        'local_validated_value': np.nan,

        'local_reference_year': np.nan,

        'validation_status': 'pdf_only_program_claim',

        'validation_note': 'No local official dataset in the repo independently reproduces this program-evaluation claim.'

    },

    {

        'indicator': 'archipelago_learning_gain_italian_months',

        'pdf_value': pdf_lookup.get('archipelago_learning_gain_italian_months'),

        'local_validated_value': np.nan,

        'local_reference_year': np.nan,

        'validation_status': 'pdf_only_program_claim',

        'validation_note': 'No local official dataset in the repo independently reproduces this program-evaluation claim.'

    },

]



validation = pd.DataFrame(validation_rows)

validation['gap_pdf_minus_local'] = validation['pdf_value'] - validation['local_validated_value']

display(validation)

validation.to_csv(PROC / 'save_the_children_validation_panel.csv', index=False)

print('Wrote:', PROC / 'save_the_children_validation_panel.csv')
```

```python
validated_only = validation[validation['validation_status'].isin(['validated_from_local_invalsi', 'partially_validated_local_official_latest_available'])].copy()
plt.figure(figsize=(10, 5))
plot_df = validated_only.dropna(subset=['local_validated_value']).copy()
plot_df['label'] = plot_df['indicator'].str.replace('_', ' ', regex=False)
x = np.arange(len(plot_df))
width = 0.38
plt.bar(x - width / 2, plot_df['pdf_value'], width=width, label='PDF extract')
plt.bar(x + width / 2, plot_df['local_validated_value'], width=width, label='Local official replication')
plt.xticks(x, plot_df['label'], rotation=20, ha='right')
plt.ylabel('Percent or months')
plt.title('Save the Children claims: PDF vs locally reproducible evidence')
plt.legend()
plt.tight_layout()
plt.show()
```
